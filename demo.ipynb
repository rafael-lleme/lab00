```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# read in data
df = pd.read_csv('dining.csv')
```

```python
# display first five rows of data
df.head()
```

```python
# plotting coffee consumed over time
sns.scatterplot(data = df, x='date', y='coffee_consumed', hue='dining_hall')
plt.xlabel('Date')
plt.ylabel('Coffee Consumed (gal)')
plt.legend(title=None)
plt.xticks(ticks=range(0, len(df['date'].unique()), 2), rotation=45)
plt.title('Dining Hall Coffee Consumption Over Time')
plt.show()
```
